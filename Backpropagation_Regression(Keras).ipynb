```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
```

```python
df
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2, activation='linear', input_dim=2))
model.add(Dense(1, activation='linear'))
```

```python
model.summary()
```

```python
# Using this we will initialize weights and biases with some random numbers,
# Exactly equal to the no of thotal trainable parameters
model.get_weights()
```

```python
# But we can initialize the weights and biases according to our need
new_weights = [np.array([[0.1, 0.1], [0.1, 0.1]], dtype=np.float32),
               np.array([0.,0.], dtype=np.float32),
               np.array([[0.1],[0.1]],dtype=np.float32),
               np.array([0.], dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error', optimizer=optimizer)
```

```python
X = df.iloc[:,0:-1].values
y = df.iloc[:,-1].values
```

```python
y
```

```python
model.fit(X,y, epochs=100, verbose=1, batch_size=1)
```
