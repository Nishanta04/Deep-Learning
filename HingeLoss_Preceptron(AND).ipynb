# **AND Logic for 2 inputs (Hinge Loss Preceptron 2D)**

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([-1,-1,-1,1])
# X = pd.DataFrame(X)
```

```python
X.shape
```

```python
y.shape
```

```python
y
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):

    w1=w2=b=1
    lr = 0.1
    count = 0

    for j in range(1000):

        for i in range(X.shape[0]):

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + b

            if z*y[i] < 0:
                count = count + 1
                w1 = w1 + lr*y[i]*X[i][0]  # update function of Gradient Decent for w1
                w2 = w2 + lr*y[i]*X[i][1]  # update function of Gradient Decent for w2
                b = b + lr*y[i]            # update function of Gradient Decent for b

    print(count)
    return w1,w2,b
```

```python
w1,w2,b = perceptron(X,y)
```

```python
y
```

```python
m = -(w1/w2)
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(-0.25,1.25,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
# plt.ylim(-3,2)
```

```python
def perceptron(X,y):

    w1=w2=w3=b=1
    lr = 0.1
    count = 0

    for j in range(1000):

        for i in range(X.shape[0]):

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + w3*X[i][2] + b

            if z*y[i] < 0:
                count = count + 1
                w1 = w1 + lr*y[i]*X[i][0]  # update function of Gradient Decent for w1
                w2 = w2 + lr*y[i]*X[i][1]  # update function of Gradient Decent for w2
                w3 = w3 + lr*y[i]*X[i][2]  # update function of Gradient Decent for w3
                b = b + lr*y[i]            # update function of Gradient Decent for b

    print(count)
    return w1,w2,w3,b
```

```python
w1,w2,w3,b = perceptron(X,y)
```

# **Hinge Loss Preceptron on Real World Data**

```python
df = pd.read_csv('placement.csv')
df
```

```python
df['placed'] = df['placed'].replace(0, -1)
display(df.head())
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(df['cgpa'],df['resume_score'],c=df['placed'],cmap='winter',s=100)
```

```python
def perceptron(df):

    X = df[['cgpa', 'resume_score']].values
    y = df['placed'].values

    w1=w2=b=1
    lr = 0.1
    count = 0

    for j in range(1000):

        for i in range(X.shape[0]):

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + b

            if z*y[i] < 0:
                count = count + 1
                # I have a doubt that the update function of the w1,w2,b should have -ve sign before y[i]
                # Like w1 = w1 + lr*-y[i]*X[i][0] and same for w2 and b

                w1 = w1 + lr*y[i]*X[i][0]  # update function of Gradient Decent for w1
                w2 = w2 + lr*y[i]*X[i][1]  # update function of Gradient Decent for w2
                b = b + lr*y[i]            # update function of Gradient Decent for b

    print(f"Number of misclassifications/updates: {count}")
    return w1,w2,b
```

```python
w1,w2,b = perceptron(df)
```

```python
m = -(w1/w2)
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(5,10,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(df['cgpa'],df['resume_score'],c=df['placed'],cmap='winter',s=100)
plt.ylim(4,10)
```
