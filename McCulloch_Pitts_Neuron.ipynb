### McCulloch-Pitts Neuron Model

```python
def mcculloch_pitts_neuron(inputs, weights, threshold):
    weighted_sum = 0
    for i in range(len(inputs)):
        weighted_sum += inputs[i] * weights[i]
    if weighted_sum >= threshold:
        return 1
    else:
        return 0
```

### Implementing AND Function

```python
and_w = [1, 1]
and_t = 2

for x1 in [0, 1]:
    for x2 in [0, 1]:
        inputs = [x1, x2]
        output = mcculloch_pitts_neuron(inputs, and_w, and_t)
        print(f"Input: {inputs}, Output (AND): {output}")
```

### Implementing OR Function

```python
or_w = [1, 1]
or_t= 1 # The neuron fires if the sum is 1 or more

for x1 in [0, 1]:
    for x2 in [0, 1]:
        inputs = [x1, x2]
        output = mcculloch_pitts_neuron(inputs, or_w, or_t)
        print(f"Input: {inputs}, Output (OR): {output}")
```

### Implementing NOT Function

```python
not_w = [-1]
not_t = 0 # If input is 0, weighted sum is 0 >= 0 -> 1. If input is 1, weighted sum is -1 < 0 -> 0

for x in [0, 1]:
    inputs = [x]
    output = mcculloch_pitts_neuron(inputs, not_w, not_t)
    print(f"Input: {inputs}, Output (NOT): {output}")
```
