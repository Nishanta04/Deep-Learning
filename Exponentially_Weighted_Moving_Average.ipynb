```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/DailyDelhiClimateTest.csv")
df = df.iloc[:,0:2]
```

```python
df
```

```python
plt.scatter(df["date"],df["meantemp"], color="red")
```

```python
x1 = df["meantemp"].ewm(alpha=0.1).mean() # alpha = (1 - beta) -> (beta = 0.9)
```

```python
df["ewma"] = x1
```

```python
df
```

```python
plt.scatter(df["date"],df["meantemp"], color="red")
plt.plot(df["date"], x1, color="black", label="beta = 0.9")
plt.legend()
```
