```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
# X consists of the feature values (Input)
# y consists the Predicted value (Output)
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
X.shape
```

```python
def perceptron(X,y):

    X = np.insert(X,0,1,axis=1)   # Inserting a new column in starting of X consisting of 1 (For bias).
    weights = np.ones(X.shape[1]) # weights[1, 1, 1] -> Initial line equation = 1 + 1*x1 + 1*x2 = 0
    lr = 0.1  # Learning Rate

    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]

    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
print(coef_)        # w1 , w2
print(intercept_)   # w0 (Bias)
```

```python
# Calculating m and b in y=mx+b
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-1,1,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
